#### 1. Secuencia de ADN dada
- **Hebra 5' $\rightarrow$ 3':** `5'- ATG CCG TTA GCT -3'`
- **Hebra 3' $\rightarrow$ 5':** `3'- TAC GGC AAT CGA -5'`

---

#### 2. Ronda de replicación (Mecanismo semiconservativo)
La replicación del ADN es **semiconservativa**: la doble hélice original se separa gracias a la helicasa y **cada una de las hebras parentales actúa como molde** para la síntesis de una nueva hebra complementaria. Por tanto, a partir de una molécula inicial se obtienen **dos moléculas hijas dobles e idénticas**:

- **Molécula Hija 1:**
  - **Hebra molde (vieja):** `5'- ATG CCG TTA GCT -3'`
  - **Hebra nueva (sintetizada):** `3'- TAC GGC AAT CGA -5'`
- **Molécula Hija 2:**
  - **Hebra molde (vieja):** `3'- TAC GGC AAT CGA -5'`
  - **Hebra nueva (sintetizada):** `5'- ATG CCG TTA GCT -3'`

##### Función de las enzimas implicadas:
- **Helicasa:** Rompe los puentes de hidrógeno entre las bases nitrogenadas, "desenrollando" y separando las dos hebras de la doble hélice para formar la horquilla de replicación.
- **Primasa:** Sintetiza un corto fragmento de ARN llamado **cebador** (o *primer*), proporcionando el extremo `3'-OH` libre indispensable para que la ADN polimerasa pueda iniciar la síntesis.
- **ADN Polimerasa:** Incorpora desoxirribonucleótidos complementarios a la hebra molde en dirección `5' -> 3'`. También posee función correctora de pruebas (*proofreading*) para corregir errores durante la síntesis.
- **Ligasa:** Sella las interrupciones en el esqueleto de fosfodiéster, uniendo los fragmentos de ADN (como los fragmentos de Okazaki de la hebra retardada).

---

#### 3. Reflexión sobre errores en la copia
Si la ADN polimerasa comete un error en la incorporación de una base y este no es corregido por los mecanismos de reparación, se produce una **mutación puntual** (sustitución). 
Al volver a replicarse la célula, esta mutación quedará fijada de forma permanente en la secuencia génica. Si la alteración ocurre dentro de una región codificante, puede cambiar el codón del ARNm y alterar el aminoácido resultante en la proteína (o generar un codón de paro prematuro), lo que podría modificar la estructura tridimensional y la función de la proteína resultante.

```python
from Bio.Seq import Seq

# 1. Secuencia de ADN dada (5' -> 3')
hebra_5_3 = Seq("ATGCCGTTAGCT")
hebra_3_5 = hebra_5_3.complement()

print("=== 1. SECUENCIA ORIGINAL (DOBLE HEBRA) ===")
print(f"Hebra 5' -> 3': {hebra_5_3}")
print(f"Hebra 3' -> 5': {hebra_3_5}")

# 2. Simulación de la Replicación Semiconservativa
molde_1 = hebra_5_3
nueva_1 = molde_1.complement()

molde_2 = hebra_3_5
nueva_2 = molde_2.complement()

print("\n=== 2. RESULTADO DE LA REPLICACIÓN (2 MOLÉCULAS HIJAS) ===")
print("Molécula Hija 1:")
print(f"  Molde (vieja, 5'->3'): {molde_1}")
print(f"  Nueva (sintetizada, 3'->5'): {nueva_1}")

print("Molécula Hija 2:")
print(f"  Molde (vieja, 3'->5'): {molde_2}")
print(f"  Nueva (sintetizada, 5'->3'): {nueva_2}")

# 3. Comprobación con Biopython
# En bioinformática es estándar representar las secuencias en sentido 5' -> 3'
print("\n=== EXTENSIÓN BIOPYTHON ===")
print(f"Complementaria Reversa (5'->3'): {hebra_5_3.reverse_complement()}")
```

## **Ejercicio 2. Transcripción del ADN a ARN**
<b>Objetivo: traducir correctamente la información de la cadena molde.
Instrucciones:
1. Usa la siguiente secuencia de ADN: 5’ – ATG CCT GAA TGC – 3’ / 3’ – TAC GGA CTT ACG – 5’.
2. Identifica cuál es la cadena molde.
3. Obtén el transcrito de ARN correspondiente (recuerda que el ARN se sintetiza en dirección 5’ →
3’ y utiliza uracilo en lugar de timina).
4. Explica cuál sería la región promotora y cuál la región codificante.

Extensión con Biopython:

#### 1. Secuencia de ADN dada
- **Hebra 5' $\rightarrow$ 3':** `5'- ATG CCT GAA TGC -3'`
- **Hebra 3' $\rightarrow$ 5':** `3'- TAC GGA CTT ACG -5'`

---

#### 2. Identificación de la cadena molde
En el proceso de transcripción, la enzima **ARN polimerasa** sintetiza el ARN siempre en dirección `5' -> 3'`. Por tanto, necesita leer el ADN en dirección `3' -> 5'`.

- **Cadena molde (antisentido):** Es la hebra `3'- TAC GGA CTT ACG -5'`. Sobre esta hebra se lee la información.
- **Cadena codificante (sentido):** Es la hebra `5'- ATG CCT GAA TGC -3'`. Su secuencia es idéntica al ARNm resultante, salvo que el ADN tiene Timina (T) y el ARN tiene Uracilo (U).

---

#### 3. Transcrito de ARN obtenido
A partir de la cadena molde (`3'- TAC GGA CTT ACG -5'`), formamos las bases complementarias de ARN (recordar: T por A, A por U, C por G y G por C):

- **ARNm transcrito (5' $\rightarrow$ 3'):** `5'- AUG CCU GAA UGC -3'`

---

#### 4. Región promotora vs. Región codificante
- **Región promotora:** Es una secuencia específica de ADN situada antes de que comience el gen (por ejemplo, la caja TATA). No se transcribe a ARN; su función es servir de señal de reconocimiento para que la ARN polimerasa y los factores de transcripción se unan al ADN e inicien el proceso.
- **Región codificante:** Es la parte del gen que realmente se transcribe a ARNm y contiene las instrucciones (codones) para sintetizar la proteína. Comienza en el sitio de inicio de transcripción y termina en la señal de terminación.

```python
from Bio.Seq import Seq

# 1. Definimos la hebra codificante (5' -> 3') y obtenemos la hebra molde (3' -> 5')
adn_codificante = Seq("ATGCCTGAATGC")
adn_molde = adn_codificante.complement()

print("=== 1. HEBRAS DE ADN ===")
print(f"Hebra Codificante (5'->3'): {adn_codificante}")
print(f"Hebra Molde       (3'->5'): {adn_molde}")

# 2. Transcripción con Biopython
# El método .transcribe() toma la hebra codificante (5'->3') y sustituye las T por U
arnm = adn_codificante.transcribe()

print("\n=== 2. TRANSCRITO DE ARN ===")
print(f"ARNm resultante   (5'->3'): {arnm}")

# 3. Experimentando al cambiar la orientación (Extensión Biopython)
# Si intentáramos transcribir desde la otra hebra (como si fuera codificante):
arnm_reverso = adn_codificante.reverse_complement().transcribe()

print("\n=== 3. EXPERIMENTO: CAMBIO DE ORIENTACIÓN ===")
print(f"ARNm si transcribimos la otra hebra: {arnm_reverso}")
```

## **Ejercicio 3. Traducción del ARNm a proteína**
<b> Objetivo: aplicar el código genético y reflexionar sobre mutaciones.
Instrucciones:
1. Usa el siguiente transcrito de ARN: 5’ – AUG UAU GCU UAA – 3’.
2. Identifica el codón de inicio y el codón de paro.
3. Traduce la secuencia en una cadena de aminoácidos.
4. Reflexiona: ¿qué pasaría si el codón de inicio mutara de AUG a GUG? ¿Qué ocurriría si el codón
de paro desapareciera por mutación?

Extensión con Biopython:
Utiliza el módulo Bio.Seq para traducir automáticamente el ARNm a una secuencia proteica y
comprueba si el resultado coincide con el tuyo. </b>

#### 1. Transcrito de ARN dado
- **Secuencia ARNm (5' $\rightarrow$ 3'):** `5'- AUG UAU GCU UAA -3'`

---

#### 2. Lectura por codones e identificación de señales
El ribosoma lee el ARNm de 3 en 3 nucleótidos (codones) en sentido `5' -> 3'`:

- **Codón de inicio:** `AUG` (Codifica para el aminoácido Metionina, Met).
- **Codones intermedios:** `UAU` (Tirosina, Tyr), `GCU` (Alanina, Ala).
- **Codón de paro (STOP):** `UAA` (No codifica ningún aminoácido; señala el fin de la traducción).

---

#### 3. Cadena de aminoácidos resultante
Traduciendo codón a codón mediante el código genético estándar:
- `AUG` $\rightarrow$ **Met** (Metionina)
- `UAU` $\rightarrow$ **Tyr** (Tirosina)
- `GCU` $\rightarrow$ **Ala** (Alanina)
- `UAA` $\rightarrow$ **STOP** (Fin del péptido)

- **Secuencia proteica / peptídica:** `Met - Tyr - Ala` (o en código de una letra: `M - Y - A`).

---

#### 4. Reflexión sobre mutaciones
- **Si el codón de inicio muta de `AUG` a `GUG`:**
  En eucariotas, el ribosoma no reconocería el punto de inicio adecuado para ensamblar el complejo de traducción. Esto provocaría que la traducción no se inicie en ese punto o que el ribosoma busque el siguiente codón `AUG` más adelante, produciendo una proteína truncada o no funcional.
- **Si el codón de paro (`UAA`) desaparece por mutación:**
  El ribosoma ignoraría este punto de parada y continuaría añadiendo aminoácidos a la cadena polipeptídica leyendo la región 3' no traducida (UTR) hasta encontrar el siguiente codón de paro en pauta. Esto genera una proteína más larga de lo normal que probablemente sufra de un mal plegamiento o pierda su función.

```python
from Bio.Seq import Seq

# 1. Definimos el transcrito de ARNm (5' -> 3')
arnm = Seq("AUGUAUGCUUAA")

print("=== 1. ARNm DE ENTRADA ===")
print(f"Secuencia de ARNm (5'->3'): {arnm}")

# 2. Traducción automática a secuencia proteica usando Biopython
# El parámetro to_stop=True detiene la traducción en cuanto encuentra el primer codón STOP
proteina = arnm.translate(to_stop=True)
proteina_con_stop = arnm.translate(to_stop=False)

print("\n=== 2. TRADUCCIÓN A PROTEÍNA ===")
print(f"Cadena de aminoácidos (hasta STOP): {proteina}")
print(f"Símbolo del codón de paro (*):       {proteina_con_stop}")

# 3. Mapeo explícito codón por aminoácido (para verificación visual)
codones = [str(arnm[i:i+3]) for i in range(0, len(arnm), 3)]
print("\n=== 3. DESGLOSE CODÓN A CODÓN ===")
for c in codones:
    sub_seq = Seq(c)
    print(f"Codón: {c} --> Aminoácido: {sub_seq.translate()}")
```
